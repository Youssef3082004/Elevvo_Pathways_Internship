```python
import json
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

## 🛠 Building Function that will Convert SQuAD Json File to CSV

```python
def convert_squad_to_csv_pandas(json_file_path:str, csv_file_path:str) -> None:
    rows = []

    with open(json_file_path, 'r', encoding='utf-8') as f:
        json_data = json.load(f)

    for article in json_data.get('data', []):
        title = article.get('title', '')
        
        for paragraph in article.get('paragraphs', []):
            context = paragraph.get('context', '')
            
            for qa in paragraph.get('qas', []):
                question_id = qa.get('id', '')
                question = qa.get('question', '')
                
                answers = qa.get('answers', [])
                if not answers:
                    rows.append({'Title': title,'Context': context,'Question_ID': question_id,'Question': question,'Answer_Text': None,'Answer_Start': None})
                else:
                    for answer in answers:
                        rows.append({'Title': title,'Context': context,'Question_ID': question_id,'Question': question,'Answer_Text': answer.get('text', ''),'Answer_Start': answer.get('answer_start', '')})

    df = pd.DataFrame(rows)
    df.to_csv(csv_file_path, index=False, encoding='utf-8')
    print(f"Successfully saved {len(df)} rows to '{csv_file_path}'")
```

```python
convert_squad_to_csv_pandas(json_file_path='Dataset/train.json',csv_file_path='train.csv')
```
