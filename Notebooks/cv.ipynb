```python
import pymupdf
import os
import warnings
import re
import string
import pandas as pd
warnings.filterwarnings("ignore")
```

```python
jobs = pd.read_csv(r"D:\My Download\Elevvo pathway\Elevvo pathway Tasks\Dataset\Jobs.csv")
companies = pd.read_csv(r"D:\My Download\Elevvo pathway\Elevvo pathway Tasks\Dataset\Companies.csv")
```

## 📍 Read Cvs from specific Title job

```python
path = "D:\My Download\Files\CV\AI Engineer"
cvs = os.listdir(path=path)
```

## 📍 Extract Text From CV

```python
cv_text = []
doc = pymupdf.open(f"{path}\{cvs[1]}") 
for i,page in enumerate(doc): 
  cv_text.append(page.get_text())

cv_text = " ".join(cv_text)
```

```python
print(cv_text)
```

## 📍 lowering Text

```python
cv_text = cv_text.lower()
```

## 📍 Removing any Special Character

```python
cv_text = re.sub(pattern=r"[^a-z\n]",string=cv_text,repl=" ")
```

## 📍 Removing any Additional Spaces

```python
cv_text = re.sub(pattern=r"\s+",string=cv_text,repl=" ")
```

## ✔ Get Matched Skills 

```python
skills = jobs[jobs["Job Title"] == "Data Scientist"]["skills"]
skills.iloc[:100]
```

```python
import spacy

def _setup_skill_extractor(knowsSkills:list[str]):
    nlp = spacy.load("en_core_web_sm")    
    ruler = nlp.add_pipe("entity_ruler", before="ner")
    patterns = [{"label": "SKILL", "pattern": [{"LOWER": skill.lower()} for skill in s.split()]} for s in knowsSkills]    
    ruler.add_patterns(patterns)
    return nlp

def extract_skills(text,knowsSkills:list[str]):
    doc = _setup_skill_extractor(knowsSkills)(text)
    found_skills = set(ent.text for ent in doc.ents if ent.label_ == "SKILL")
    return list(found_skills)

known_skills = skills.iloc[0:].apply(lambda x: re.findall(r"\b[A-Z]+[a-z]*(?:\s+[a-z]+)*",string=x)).to_list()[0]
    
cv_skills = extract_skills(cv_text, known_skills)
    
print("Skills Found in CV:")
print(cv_skills)
```
