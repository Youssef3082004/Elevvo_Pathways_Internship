```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pymupdf
import os
import warnings 
warnings.filterwarnings("ignore")
```

## 📍 Lodaing **`all-MiniLM-L6-v2`** Model

```python
model = SentenceTransformer(model_name_or_path="all-MiniLM-L6-v2")
```

## 📍 Load All Resumes from Folder

```python
path = r"D:/My Download/Elevvo pathway/Elevvo pathway Tasks/Dataset/Resumes/HR"
cvs = os.listdir(path)
```

## 📍 Extract Text from Each Resume

```python
files = []
for cv in cvs:
    doc = pymupdf.open(filename=fr"{path}/{cv}")
    file_content = []

    for page in doc:

        file_content.append(page.get_text())
    files.append(" ".join( file_content))
```

```python
print(f"Number of Resumes: {len(files)}")
```

## 🚩 HR Job Description

```python
job_description = """
Job Title: Human Resources (HR) Specialist

Job Summary:
We are seeking a proactive and organized Human Resources (HR) Specialist to support our organization in recruitment, employee relations, performance management, and HR operations. The ideal candidate will help build a positive workplace culture while ensuring compliance with company policies and labor regulations.

Key Responsibilities:
Manage the full recruitment cycle, including job posting, screening, interviewing, and onboarding.
Develop and implement HR policies and procedures.
Maintain accurate employee records and HR documentation.
Support performance management and employee evaluation processes.
Handle employee relations matters professionally and confidentially.
Coordinate training and development programs.
Assist in payroll preparation and benefits administration.
Ensure compliance with labor laws and company regulations.
Prepare HR reports and provide insights to management.

Qualifications:
Bachelor’s degree in Human Resources, Business Administration, or a related field.
1–3 years of experience in HR or a similar role.
Knowledge of HR practices and labor regulations.
Strong communication and interpersonal skills.
Excellent organizational and problem-solving abilities.
Proficiency in Microsoft Office and HR management systems.
Ability to handle confidential information with integrity.

Preferred Skills:
Experience with HRIS systems.
Familiarity with recruitment platforms and applicant tracking systems.
Analytical and reporting skills.
Ability to work effectively in a fast-paced environment.

Benefits:
Competitive salary package.
Professional development opportunities.
Supportive and collaborative work environment.
Opportunities for career growth.
"""
```

## 🟢 Encoding Resumes Text & Job Description 

```python
cvs_embedding = model.encode(files)
job_description_embedding = model.encode([job_description])
```

## 🟢 Calculate Cosine Similarity
* This returns a score between **0 and 1** representing semantic closeness

```python
similarities = cosine_similarity(job_description_embedding,cvs_embedding)[0]
```

## 🟢 Mapping Scores with Resumes 

```python
scores = {}
for file,score in zip(cvs,similarities):
    scores[file] = round(score *100 ,2)
```

```python
sorted(scores.items(),key=lambda x:x[1],reverse=True)
```

## 📍 Functions

```python
def _Read_CVS(path:str) -> tuple[list[str],list[str]]:
    cvs = os.listdir(rf"{path}")
    files = []
    for cv in cvs:
        doc = pymupdf.open(filename=fr"{path}/{cv}")
        file_content = []

        for page in doc:

            file_content.append(page.get_text())
        files.append(" ".join( file_content))
    return files ,cvs
```

```python
def GetTopCVs(job_description:str,path:str) -> list[tuple[str,float]]:
    CVFiles ,cvs_names = _Read_CVS(path=path)
    model = SentenceTransformer(model_name_or_path="all-MiniLM-L6-v2")
    cvs_embedding = model.encode(CVFiles)
    job_description_embedding = model.encode([job_description])
    similarities = cosine_similarity(job_description_embedding,cvs_embedding)[0]

    scores = {}
    for file,score in zip(cvs_names,similarities):
        scores[file] = round(score *100 ,2)
    return sorted(scores.items(),key=lambda x:x[1],reverse=True)
```

```python
GetTopCVs(job_description=job_description,path=path)
```
