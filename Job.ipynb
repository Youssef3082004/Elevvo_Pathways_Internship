## ⚠ Splitting Dataset 
* We should split data becauce we have a lot of Job Descriptions 
* we will split data to 2 Datasets
* first one for **Companies** and Second one for **Jobs**

```python
jobs.columns
```

## ✅ Here we Maked **Companies** Dataset

```python
CompanyColumns = ["Job Id",'Company', 'Company Profile',"Company Size","location","Country", 'latitude', 'longitude','Contact Person', 'Contact']
companies = jobs.loc[:,CompanyColumns]
companies.to_csv("Dataset/Companies.csv",index=False)
```

## 🟢 Now we will make **Jobs** Dataset

```python
columns = [column for column in jobs.columns.to_list() if column not in CompanyColumns]
columns.append("Job Id")
```

## 📍 Here we dropped any column in **Companies** Dataset  

```python
jobs = jobs.loc[:,columns]
```

## 🟢 Converting **`Job Posting Date`** to Datetme Format 

```python
new_column = pd.to_datetime(jobs["Job Posting Date"],format="%Y-%m-%d")
```

```python
jobs["Date"] = new_column
```

```python
jobs.drop(columns=["Job Posting Date"],inplace=True)
jobs.head()
```

```python
np.unique(jobs["Job Title"])
```

```python
jobs.to_csv("Dataset/Jobs.csv",index=False)
```
